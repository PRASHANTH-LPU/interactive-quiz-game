```python

import pygame
import sys

# Initialize Pygame
pygame.init()

# Screen settings
WIDTH, HEIGHT = 800, 500
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Interactive Quiz Game")

# Colors
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
BLUE = (50, 120, 220)
GREEN = (50, 180, 80)
RED = (220, 60, 60)
GRAY = (220, 220, 220)

# Fonts
font = pygame.font.Font(None, 36)
big_font = pygame.font.Font(None, 50)

# Quiz questions
questions = [
    {
        "question": "Which language is used for this project?",
        "options": ["Java", "Python", "C++", "HTML"],
        "answer": "Python"
    },
    {
        "question": "Which library is used for graphics?",
        "options": ["NumPy", "Pandas", "Pygame", "Tkinter"],
        "answer": "Pygame"
    },
    {
        "question": "Which keyword is used to define a function?",
        "options": ["function", "define", "def", "fun"],
        "answer": "def"
    },
    {
        "question": "Which symbol starts a comment in Python?",
        "options": ["//", "#", "/*", "&"],
        "answer": "#"
    },
    {
        "question": "Which function displays output in Python?",
        "options": ["input()", "print()", "show()", "display()"],
        "answer": "print()"
    }
]

score = 0
current_question = 0
selected = False
result = ""

# Draw text
def draw_text(text, font, color, x, y):
    image = font.render(text, True, color)
    screen.blit(image, (x, y))

# Main game loop
running = True

while running:
    screen.fill(WHITE)

    for event in pygame.event.get():

        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.MOUSEBUTTONDOWN:

            if current_question < len(questions) and not selected:

                mouse_pos = pygame.mouse.get_pos()

                for i in range(4):
                    button = pygame.Rect(150, 180 + i * 65, 500, 50)

                    if button.collidepoint(mouse_pos):

                        answer = questions[current_question]["options"][i]

                        if answer == questions[current_question]["answer"]:
                            score += 1
                            result = "Correct Answer!"
                        else:
                            result = "Wrong Answer!"

                        selected = True

            elif selected:
                current_question += 1
                selected = False

    # Display quiz
    if current_question < len(questions):

        draw_text("INTERACTIVE QUIZ GAME", big_font, BLUE, 180, 30)

        q = questions[current_question]

        draw_text(
            f"Question {current_question + 1}: {q['question']}",
            font, BLACK, 80, 110
        )

        for i, option in enumerate(q["options"]):

            button = pygame.Rect(150, 180 + i * 65, 500, 50)

            pygame.draw.rect(screen, BLUE, button, border_radius=8)

            draw_text(option, font, WHITE, 170, 192 + i * 65)

        if selected:
            color = GREEN if result == "Correct Answer!" else RED
            draw_text(result, font, color, 300, 450)
            draw_text("Click anywhere for next question", font, BLACK, 190, 470)

    else:

        draw_text("QUIZ COMPLETED!", big_font, GREEN, 230, 150)

        draw_text(
            f"Your Score: {score}/{len(questions)}",
            font, BLACK, 300, 230
        )

        draw_text("Thank you for playing!", font, BLUE, 270, 300)

    pygame.display.update()

pygame.quit()
sys.exit()
```
